### Q3. Write a function that prints the digits of a number n. For eg: n = 312, there are 3 digits in it 3, 1 and 2 & we need to print them.

- Hint- The right most digit of a number N is N%10. And to remove the right most digit from a number, we can do N = N / 10.

```python
def print_digits_math(n):
    # Handle negative numbers
    n = abs(n)

    # Handle 0 explicitly
    if n == 0:
        print(0)
        return

    # Helper function to print digits left-to-right using recursion
    def helper(num):
        if num == 0:
            return
        helper(num // 10)  # Step 1: Remove rightmost digit and recurse
        print(num % 10)  # Step 2: Print current rightmost digit on unwinding

    helper(n)


# Example usage:
print_digits_math(312)
```

### Q7. Design a program to continuously input a number n from user & print if it is positive or negative until the user enters "Quit".

```python
while True:
    user_input = input("Enter a number or 'Quit': ").title()

    if user_input == "Quit":
        print("Program closed.")
        break

    try:
        n = float(user_input)
        if n > 0:
            print(f"{n} is Positive")
        elif n < 0:
            print(f"{n} is Negative")
        else:
            print("The number is Zero")
    except ValueError:
        print("Invalid input! Please enter a valid number or 'Quit'.")
```

### Q9. Write a function is_p rime (n) that returns True if n is a prime number and False otherwise, using a loop.

```python
def is_prime(n):
    if n <= 1:
        return False

    # Check for factors from 2 up to the square root of n
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False  # Found a divisor, so it's not prime

    return True  # No divisors found, it is prime


# Example testing:
print(is_prime(4))  # Returns False
print(is_prime(7))  # Returns True
```
