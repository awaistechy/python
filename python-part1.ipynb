### Q2 Take two numbers as input from the user and print their sum, difference, product, and quotient.

```python
# Task 2
num1 = int(input("Enter 1st Number "))
num2 = int(input("Enter 2nd Number "))
sum = num1 + num2
difference = num1 - num2
product = num1 * num2
quotient = num1 / num2
print(f"sum of {num1} and {num2} is {sum}")
print(f"difference of {num1} and {num2} is {difference}")
print(f"product of {num1} and {num2} is {product}")
print(f"quotient of {num1} and {num2} is {quotient}")  # Type Conversion
```

### Q3 Ask the user to enter two integers and one float. Convert them all to floats and print their average.

```python
# Task 3
num1 = float(input("Enter an integer value "))
num2 = float(input("Enter a second integer value "))
num3 = float(input("Enter a float value "))
sum = num1 + num2 + num3
average = sum / 3
print(f"Average of {num1}, {num2} and {num3} is {average}")
```

### Q4 The user enters a string containing a number (e.g., "45" ). Convert it to: an integer, a float, a string again Print all three values with their types.

```python
# Task 4
number = input("Enter a number ")
integer_value = int(number)
float_value = float(number)
string_value = str(number)
print(f"Integer {integer_value} type of {type(integer_value)}")
print(f"Float {float_value} type of {type(float_value)}")
print(f"String {string_value} type of {type(string_value)}")
```

### Q6 Write a program to swap values of two numbers entered by the user.

```python
# Task 6
num1 = int(input("Enter 1st number "))
num2 = int(input("Enter 2nd number "))
print(f"Value of number1 is {num1} and number2 is {num2} before Swaping")
temp = num1
num1 = num2
num2 = temp
print(f"Value of number1 is {num1} and number2 is {num2} after Swaping")
```

### Q7  Ask the user for a temperature in Celsius (string input). Convert it to float , then calculate and print temperature in Fahrenheit.
Conversion formula: Fahrenheit = (Celsius * 9/5) + 32

```python
# Task 7
temp_in_celsius = float(input("Enter temperature in celsius "))
temp_in_fahrenheit = (temp_in_celsius * 9 / 5) + 32
print(f"Temperature in Fahrenheit is {temp_in_fahrenheit}")
```

### Q8 Take the radius (r) as user input and print the area.
Use the formula: Area = π * r ** 2   
(value of π = 3.14)

```python
# Task 8
radius = float(input("Enter radius of circle "))
area = 3.14 * radius**2
print(f"Area = {area}")
```

### Q9 Ask the user for:  Principal (P), Rate (R), Time (T). Convert all to float and compute simple interest: 
SI =(P ∗R∗T)/100

```python
# Task 9
principal = float(input("Enter principal amount "))
rate = float(input("Enter rate "))
time = float(input("Enter time "))
simple_interest = (principal * rate * time) / 100
print(f"Simple Interest = {simple_interest}")
```

### Q10 Take a decimal number as input (like 45.78) and output its:
- integer part - 45
- fractional part - .78

```python
# Task 10
decimal_number = float(input("Enter a decimal number "))
integer_number = int(decimal_number)
fraction = round(decimal_number - integer_number,2)
print(f"Integer part is {integer_number} and fraction part is {fraction}")
```
